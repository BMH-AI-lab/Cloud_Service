```python
def solution(n):
    answer = 0
    m = 1

    while m * (m - 1) // 2 < n:
        if (n - m * (m - 1) // 2) % m == 0:
            answer += 1
        m += 1

    return answer

solution(15)
```

```python
def solution(n):
    answer = n + 1
    zero = bin(n)[2:].count('1')
    while(zero != bin(answer)[2:].count('1')):
        answer += 1
    return answer
solution(78)
```

```python
def solution(s):
    answer = 0
    stack = []
    for char in s:
        if stack and stack[-1] == char:
            stack.pop()
        else:
            stack.append(char)
    if len(stack) == 0:
        answer = 1
    return answer

solution("baabaa")
```

```python
def solution(n):
    a = [0, 1] + [0] * (n - 1)
    for i in range(2, n + 1):
        a[i] = (a[i - 1] + a[i - 2]) % 1234567
    return a[n]
solution(5)
```
