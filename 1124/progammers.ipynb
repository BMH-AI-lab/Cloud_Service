```python
def solution(people, limit):
    people.sort()
    left, right = 0, len(people) - 1
    boats = 0

    while left <= right:
        if people[left] + people[right] <= limit:
            left += 1
        right -= 1
        boats += 1

    return boats

solution([70, 50, 80, 50], 100) 
```

```python
def solution(k, tangerines):
    from collections import Counter

    count = Counter(tangerines)
    frequencies = sorted(count.values(), reverse=True)

    total = 0
    types = 0

    for freq in frequencies:
        total += freq
        types += 1
        if total >= k:
            return types

    return types    

solution(4, [1, 3, 2, 5, 4, 5, 2, 3])
```

```python
def solution(n):
    MOD = 123456789
    dp = [0] * (n + 1)
    dp[0], dp[1] = 1, 1

    for i in range(2, n + 1):
        dp[i] = (dp[i - 1] + dp[i - 2]) % MOD
    return dp[n]

solution(4) 
```
